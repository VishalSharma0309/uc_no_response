```python
# Standard setup for Kedro + Jupyter integration
import os
from pathlib import Path
from kedro.framework.startup import bootstrap_project
from kedro.framework.session import KedroSession
from kedro.framework.project import configure_project

# 1. Find the project root (assuming notebook is in notebooks/ folder)
project_path = Path.cwd().parent  # Adjust if your notebook is deeper
print(project_path)

# 2. Bootstrap the Kedro project
# if not (project_path / ".kedro.yml").exists():
#     raise ValueError(f"Kedro project not found at {project_path}")

bootstrap_project(project_path)
configure_project(project_path.name)

# 3. Create a session and load the catalog
session = KedroSession.create(project_path=project_path)
context = session.load_context()

# 4. Make catalog available
catalog = context.catalog
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
engineered_data = catalog.load("data_with_new_features")  # From your feature engineering node
target_col = catalog.load("params:target_column")
analysis_features = catalog.load("params:data_preprocessing.engineered_features_analysis")
```

```python
# Feature Summary Statistics

def describe_engineered_features(df):
    """Generate custom stats for engineered features"""
    
    stats = pd.DataFrame({
        'dtype': df[analysis_features].dtypes,
        'missing_pct': df[analysis_features].isna().mean(),
        'zero_pct': (df[analysis_features] == 0).mean(),
        'n_unique': df[analysis_features].nunique()
    })
    
    # Add numerical stats
    num_cols = df[analysis_features].select_dtypes(include=np.number).columns
    stats = stats.join(df[num_cols].agg(['mean', 'median', 'std', 'skew']).T)
    
    return stats

feature_stats = describe_engineered_features(engineered_data)
display(feature_stats.style.background_gradient(cmap='Blues'))
```

```python
# Target Relationship Analysis

plt.figure(figsize=(14, 10))
for i, col in enumerate(analysis_features, 1):
    plt.subplot(2, 3, i)
    if engineered_data[col].nunique() > 10:
        sns.boxplot(x=target_col, y=col, data=engineered_data)
    else:
        sns.countplot(x=col, hue=target_col, data=engineered_data)
    plt.title(f"{col} vs {target_col}")
    plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# Correlation Analysis
numeric_data = engineered_data.select_dtypes(include=np.number)

# Correlation matrix
plt.figure(figsize=(12, 10))
corr_matrix = numeric_data.corr()
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap='coolwarm', 
            center=0, vmin=-1, vmax=1, mask=np.triu(np.ones_like(corr_matrix)))
plt.title("Feature Correlation Matrix")
plt.show()

# Top correlations with target
target_corr = corr_matrix[target_col].sort_values(ascending=False)
display(target_corr.to_frame("Correlation with Target").style.background_gradient(cmap='coolwarm'))
```

```python
sns.lmplot(
    x='delivery_intensity',
    y='NR_frequency',
    hue=target_col,
    data=engineered_data,
    height=6,
    aspect=1.5,
    logistic=True if engineered_data[target_col].nunique() == 2 else False
)
plt.title("Interaction: Delivery Intensity vs NR Frequency")
plt.show()
```
